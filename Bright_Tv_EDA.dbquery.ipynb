```sql
--I wanted to check the whole table before i start doing any analysis
select *
from brighttvanalysis.default._bright_tv_user_profile
limit 10;
------------------------------------------------------------
-- checking for duplicates in my data
select UserID, COUNT(*) AS  duplicate_count
from brighttvanalysis.default._bright_tv_user_profile
GROUP BY UserID
HAVING count(*) > 1; 


-- how big is this data
select COUNT(*) AS number_of_rows,
COUNT(DISTINCT UserID) AS number_of_subs
from brighttvanalysis.default._bright_tv_user_profile;
-- the number of rows equals  to the number of distinct users meaning that there is no duplicate 

--are there any rows where user id is null
select COUNT(*) AS cnt
from brighttvanalysis.default._bright_tv_user_profile
WHERE UserID IS NULL;

select*
from brighttvanalysis.default._bright_tv_user_profile;

------------------------------------------------------------

select DISTINCT gender 
from brighttvanalysis.default._bright_tv_user_profile;


-- checkimg how many have an empty space on the gender
select COUNT(*)
from brighttvanalysis.default._bright_tv_user_profile
WHERE gender=' ';

select count(*) AS cnt, 
count(DISTINCT userid )  AS subs, 
     CASE 
         WHEN gender = ' ' THEN 'unknown'
          WHEN gender ilike '%none%' THEN 'unknown' 
          ELSE gender
        END AS gender  
from brighttvanalysis.default._bright_tv_user_profile
GROUP BY gender;
--------------------------------------------------------------
--------------------------------------------------------------
select DISTINCT Race  
from brighttvanalysis.default._bright_tv_user_profile;

-- checking if there is a row where race is null
select count (*) AS num_of_rows
from brighttvanalysis.default._bright_tv_user_profile
WHERE Race IS NULL;

--- below i have combined none, other and empty space into one thing 'None'
 select DISTINCT
      CASE 
          WHEN Race='other' THEN 'None'
          WHEN Race=' ' THEN 'None'
        ELSE Race
    END AS Race
 from brighttvanalysis.default._bright_tv_user_profile;  
---------------------------------------------------------------
-- province checks
--------------------------------------------------------------

Select DISTINCT  province
from brighttvanalysis.default._bright_tv_user_profile; 

Select DISTINCT 
            CASE 
                WHEN province=' ' THEN 'uncategorized'
                WHEN province='None' THEN 'uncategorized'
                ELSE province
            END AS province
from brighttvanalysis.default._bright_tv_user_profile; 
---------------------------------------------------------------
-- age checks
-------------------------------------------------------------- 

select MIN(Age) AS min_age, ---=0
       MAX(Age) AS max_age ---=114
from brighttvanalysis.default._bright_tv_user_profile; 

Select Count(*) AS cnt
from brighttvanalysis.default._bright_tv_user_profile
WHERE Age IS NULL;
--Age buckets
select COUNT(DISTINCT userid) As subs,
      CASE 
          WHEN age = 0 THEN 'Infants'
          WHEN age BETWEEN 1 AND 12 THEN 'Kids'
          WHEN age BETWEEN 13 AND 19 THEN 'Teenager'
          WHEN age BETWEEN 20 AND 35 THEN 'Youth'
          WHEN age BETWEEN 36 AND 50 THEN 'Adults'
          WHEN age BETWEEN 51 AND 65 THEN 'Elder'
          WHEN age >=65 THEN 'Senior'
        END AS age_groups
from brighttvanalysis.default._bright_tv_user_profile
GROUP BY age_groups;


-------------------------------------------------------------------------------------
-- putting the code together ( the casea or everything that have been created)
-------------------------------------------------------------------------------------


with cte1 AS (
    select UserID,

          CASE 
                WHEN province=' ' THEN 'uncategorized'
                WHEN province='None' THEN 'uncategorized'
                ELSE province
            END AS province,

            age,
        CASE 
          WHEN age = 0 THEN 'Infants'
          WHEN age BETWEEN 1 AND 12 THEN 'Kids'
          WHEN age BETWEEN 13 AND 19 THEN 'Teenager'
          WHEN age BETWEEN 20 AND 35 THEN 'Youth'
          WHEN age BETWEEN 36 AND 50 THEN 'Adults'
          WHEN age BETWEEN 51 AND 65 THEN 'Elder'
          WHEN age >=65 THEN 'Senior'
        END AS age_groups, 

        --creating a flag of social media and an email using CASE
        CASE
           
            WHEN (Email IS NOT NULL) OR ( Email=' ') OR (Email NOT IN ('None')) THEN 1
           ELSE 0
           END AS email_flag,
        CASE 
            WHEN `Social Media Handle` IS NOT NULL OR `Social Media Handle`=' ' OR `Social Media Handle` NOT IN ( 'None') THEN 1
            ELSE 0
        END AS social_media_flag,
        ----------------
      CASE 
          WHEN Race='other' THEN 'None'
          WHEN Race=' ' THEN 'None'
        ELSE Race
    END AS Race,

    CASE 
         WHEN gender = ' ' THEN 'unknown'
          WHEN gender ilike '%none%' THEN 'unknown' 
          ELSE gender
        END AS gender 

    from brighttvanalysis.default._bright_tv_user_profile 

)  
    Select* 
    from cte1;

----------------------------------------------------------------------------------------------------















```
